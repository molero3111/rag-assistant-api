## Authentication Workflow

Let's walk through registration, login, token refresh, and authenticated API usage. We'll store the access token for use in profile and chat requests.

````python
import requests

API_BASE = "http://localhost:8001/api/"

# Registration (adjust email/username for repeated runs)
register_payload = {
    "username": "testuser1",
    "email": "testuser1@example.com",
    "password": "TestPassword123!"
}
registration_response = requests.post(API_BASE + "register/", json=register_payload)
registration_data = registration_response.json()
access_token = registration_data.get("access")
refresh_token = registration_data.get("refresh")
print("Registration:", registration_data)
print(registration_response.json())
````

````python
# Login
login_payload = {
    "username": "testuser1",
    "password": "TestPassword123!"
}
login_response = requests.post(API_BASE + "login/", json=login_payload)
print("Login:", login_response)
login_data = login_response.json()
print(login_data)

access_token = login_data.get("access")
refresh_token = login_data.get("refresh")
print("Access token:", access_token)
print("Refresh token:", refresh_token)
````

````python
# Refresh access token
refresh_payload = {"refresh": refresh_token}
refresh_response = requests.post(API_BASE + "token/refresh/", json=refresh_payload)
print("Refresh status:", refresh_response.status_code)
refresh_data = refresh_response.json()
print(refresh_data)

if "access" in refresh_data:
    access_token = refresh_data["access"]  # Update access token if refreshed
    print("New access token:", access_token)
````

````python
# Get user profile (JWT required)
headers = {"Authorization": f"Bearer {access_token}"}
profile_response = requests.get(API_BASE + "profile/", headers=headers)
print("Profile status:", profile_response.status_code)
print(profile_response.json())
````

## Send a Prompt to the RAG Chat Endpoint (Authenticated)

Now let's send a prompt to the chat endpoint using the Bearer token.

````python
# Send a prompt to the chat endpoint (JWT required)
chat_payload = {
    "prompt": "I'm stuck on this equation: 3(2x - 4) = 5x + 6. Can you help me solve for x and explain the steps?"
}
chat_headers = {"Authorization": f"Bearer {access_token}"}
chat_response = requests.post(
    API_BASE + "rag/chat/",
    json=chat_payload,
    headers=chat_headers,
    timeout=120  # wait up to 120 seconds
)
print("Chat status:", chat_response.status_code)
print(chat_response.json())
````

## Animals Endpoints: Structured and Model-Native Output
The API provides two endpoints for animal lists:
- `/api/rag/animals-list/` — Structured output (not model-native)
- `/api/rag/animals/` — Model-native structured output (requires DeepSeek API key)

**Note:** For `/api/rag/animals/`, you must set your DeepSeek API key in `.env`:
```env
LLM_MODEL=deepseek-chat
LLM_API_KEY=your_deepseek_api_key_here
```

````python
# Example: Structured animals list (not model-native)
animals_list_payload = {
    "prompt": "List 5 animals that are herbivores and green in color."
}
animals_list_response = requests.post(
    API_BASE + "rag/animals-list/",
    json=animals_list_payload,
    headers=chat_headers,
    timeout=600
 )
print("Structured animals list:", animals_list_response.status_code)
print(animals_list_response.json())

# Example: Model-native structured animals list (requires DeepSeek API key)
animals_native_payload = {
    "prompt": "List 5 animals that are herbivores and green in color."
}
animals_native_response = requests.post(
    API_BASE + "rag/animals/",
    json=animals_native_payload,
    headers=chat_headers,
    timeout=600
 )
print("Model-native animals list:", animals_native_response.status_code)
print(animals_native_response.json())
````
